```python
def get_lat_long(address):
    address = address + ", Ghana"
    location = nom.geocode(address)
    if location:
        return (location.latitude, location.longitude)
    else:
        return None
        
```

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import StandardScaler

# Selecting relevant features
content_features = df[['Avg Rating', 'Avg Price', 'Upper']].copy()

content_features.dropna(inplace=True)

scaler = StandardScaler()
content_features_scaled = scaler.fit_transform(content_features)

similarity_matrix = cosine_similarity(content_features_scaled)


def recommend_similar_restaurants(restaurant_name, top_n=5):
    # Get index of the input restaurant
    idx = df[df['Restaurant Name'] == restaurant_name].index[0]
    
    # Get similarity scores for the restaurant
    similarity_scores = list(enumerate(similarity_matrix[idx]))
    
    # Sort restaurants based on similarity scores
    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)
    
    # Get the top_n most similar restaurants
    top_similar_restaurants = similarity_scores[1:top_n+1]
    
    # Return the restaurant names
    similar_restaurants = [df.iloc[i[0]]['Restaurant Name'] for i in top_similar_restaurants]
    return similar_restaurants


```

```python
similar_restaurants = recommend_similar_restaurants('Sub Box(The Sandwich & Grill Shop)')
print(similar_restaurants)
```

```python
df['Coordinates'] = df['Location'].apply(get_lat_long)
```

```python
def filter_by_location(user_lat, user_lon, max_distance_km=12):
    user_coords = (user_lat, user_lon)
    
    
    df['Distance_km'] = df['Coordinates'].apply(lambda x: geodesic(user_coords, x).km)
    
    nearby_restaurants = df[df['Distance_km'] <= max_distance_km]

    return nearby_restaurants
```

### Two versions of the recommender system
##### - The first one requires an input restaurant to be able to fetch similar restaurants
##### - The second one just recommends restaurants for a user based on proximity,rating and popularity 

```python
def pasara_recommender1(user_lat, user_lon, restaurant_name, max_distance_km=12, top_n=10):
    nearby_restaurants = filter_by_location(user_lat, user_lon, max_distance_km)
    
    if nearby_restaurants.empty:
        return "No nearby restaurants found."
    
    nearby_indices = nearby_restaurants.index
    
    try:
        input_idx =df[df['Restaurant Name'] == restaurant_name].index[0]
    except IndexError:
        return "Restaurant not found."
    
    similarity_scores = list(enumerate(similarity_matrix[input_idx]))
    
    # Filter similarity scores to only include nearby restaurants
    similarity_scores = [(i, score) for i, score in similarity_scores if i in nearby_indices]
    
    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)
    
    top_similar_nearby = similarity_scores[1:top_n+1]
    
    recommended_restaurants = [df.iloc[i[0]]['Restaurant Name'] for i in top_similar_nearby]
    
    return recommended_restaurants


```

```python

recommendations = pasara_recommender1(5.6037, -0.1870, 'Sub Box(The Sandwich & Grill Shop)', 5, 5)
print(recommendations)
```

```python
def pasara_recommender2(user_lat, user_lon, max_distance_km=12, top_n=10):
    
    nearby_restaurants = filter_by_location(user_lat, user_lon, max_distance_km)
    
    if nearby_restaurants.empty:
        return "No nearby restaurants found."
    

    nearby_restaurants = nearby_restaurants.sort_values(by=['Rating Count','Avg Rating'], ascending=[False, False])
    

    top_recommended = nearby_restaurants.head(top_n)
    
    recommended_restaurants = top_recommended['Restaurant Name'].tolist()
    
    return recommended_restaurants


```

```python
recommendations = pasara_recommender2(5.6037, -0.1870, 5, 5)
print(recommendations)
```
